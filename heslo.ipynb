```python
import secrets
import string

def generate_password(length=32):
    uppercase = ''.join(c for c in string.ascii_uppercase if c != 'O')
    lowercase = ''.join(c for c in string.ascii_lowercase if c != 'l')
    digits = '23456789'

    allowed_chars = uppercase + lowercase + digits
    password = ''.join(secrets.choice(allowed_chars) for _ in range(length))
    return password

# Example usage:
for _ in range(100):
  print(generate_password())
```
